```python
import numpy as np
import pandas as pd
import random
from datetime import datetime, timedelta
from simglucose.simulation.env import T1DSimEnv
from simglucose.controller.basal_bolus_ctrller import BBController
from simglucose.sensor.cgm import CGMSensor
from simglucose.actuator.pump import InsulinPump
from simglucose.patient.t1dpatient import T1DPatient
from simglucose.simulation.scenario import CustomScenario
from simglucose.simulation.sim_engine import SimObj, sim, batch_sim

seed = 42
np.random.seed(seed)
random.seed(seed)

start_time = datetime(2026, 10, 24, 0, 0, 0)
patient_name = 'adolescent#001'

scenario = CustomScenario(start_time=start_time, scenario=[
    (timedelta(hours=8), 60),
    (timedelta(hours=12), 80),
    (timedelta(hours=18), 70)
])

patient = T1DPatient.withName(patient_name)
sensor = CGMSensor.withName('Dexcom', seed=seed)
pump = InsulinPump.withName('Insulet')

env = T1DSimEnv(patient, sensor, pump, scenario)
controller = BBController()

sim_instance = SimObj(env, controller, timedelta(days=14), animate=False, path='./results')
results = sim(sim_instance)

df = results.copy()
df = df[['CGM']]
df.reset_index(inplace=True)
df.rename(columns={'index': 'Time', 'CGM': 'Glucose_mg_dL'}, inplace=True)

print("In-silico dataset generated successfully.")
print(df.head())
```

```python
import torch
from torch.utils.data import Dataset, DataLoader
from sklearn.preprocessing import StandardScaler
import numpy as np

scaler = StandardScaler()
glucose_scaled = scaler.fit_transform(df[['Glucose_mg_dL']].values)

def create_sliding_windows(data, seq_length, pred_length):
    x, y = [], []
    for i in range(len(data) - seq_length - pred_length + 1):
        x.append(data[i : i + seq_length])
        y.append(data[i + seq_length : i + seq_length + pred_length])
    return np.array(x), np.array(y)

seq_length = 96
pred_length = 24

x_data, y_data = create_sliding_windows(glucose_scaled, seq_length, pred_length)

class CGMDataset(Dataset):
    def __init__(self, x, y):
        self.x = torch.tensor(x, dtype=torch.float32)
        self.y = torch.tensor(y, dtype=torch.float32)

    def __len__(self):
        return len(self.x)

    def __getitem__(self, idx):
        return self.x[idx], self.y[idx]

dataset = CGMDataset(x_data, y_data)
dataloader = DataLoader(dataset, batch_size=32, shuffle=True)

x_batch, y_batch = next(iter(dataloader))
print(f"Total sliding windows created: {len(dataset)}")
print(f"Input shape (Batch, Seq_Len, Features): {x_batch.shape}")
print(f"Target shape (Batch, Pred_Len, Features): {y_batch.shape}")
```

```python
import torch
import torch.nn as nn

class DualPenaltyLoss(nn.Module):
    def __init__(self, alpha=1.0, beta=5.0):
        super(DualPenaltyLoss, self).__init__()
        self.alpha = alpha
        self.beta = beta

    def forward(self, y_pred, y_true):
        error_sq = (y_true - y_pred) ** 2

        mask_alpha = (y_pred > y_true).float()
        mask_beta = (y_pred <= y_true).float()

        loss = (self.alpha * error_sq * mask_alpha) + (self.beta * error_sq * mask_beta)

        return torch.mean(loss)

criterion = DualPenaltyLoss(alpha=1.0, beta=5.0)

dummy_predictions = y_batch + (torch.randn_like(y_batch) * 0.5)

loss_value = criterion(dummy_predictions, y_batch)
print(f"Dual-Penalty Loss for the batch: {loss_value.item():.4f}")
```

```python
import torch
import torch.nn as nn

class moving_avg(nn.Module):
    def __init__(self, kernel_size, stride):
        super(moving_avg, self).__init__()
        self.kernel_size = kernel_size
        self.avg = nn.AvgPool1d(kernel_size=kernel_size, stride=stride, padding=0)

    def forward(self, x):
        front = x[:, 0:1, :].repeat(1, (self.kernel_size - 1) // 2, 1)
        end = x[:, -1:, :].repeat(1, (self.kernel_size - 1) // 2, 1)
        x = torch.cat([front, x, end], dim=1)
        x = self.avg(x.permute(0, 2, 1))
        x = x.permute(0, 2, 1)
        return x

class series_decomp(nn.Module):
    def __init__(self, kernel_size):
        super(series_decomp, self).__init__()
        self.moving_avg = moving_avg(kernel_size, stride=1)

    def forward(self, x):
        moving_mean = self.moving_avg(x)
        res = x - moving_mean
        return res, moving_mean

class DLinear(nn.Module):
    def __init__(self, seq_len, pred_len):
        super(DLinear, self).__init__()
        self.seq_len = seq_len
        self.pred_len = pred_len
        self.decomposition = series_decomp(25)
        self.Linear_Seasonal = nn.Linear(self.seq_len, self.pred_len)
        self.Linear_Trend = nn.Linear(self.seq_len, self.pred_len)

    def forward(self, x):
        seasonal_init, trend_init = self.decomposition(x)
        seasonal_init = seasonal_init.permute(0, 2, 1)
        trend_init = trend_init.permute(0, 2, 1)

        seasonal_output = self.Linear_Seasonal(seasonal_init)
        trend_output = self.Linear_Trend(trend_init)

        x = seasonal_output + trend_output
        return x.permute(0, 2, 1)

model_dlinear = DLinear(seq_len=96, pred_len=24)
sample_output = model_dlinear(x_batch)

print(f"DLinear Output Shape: {sample_output.shape}")
```

```python
import torch.optim as optim

model_dlinear = DLinear(seq_len=96, pred_len=24)
criterion = DualPenaltyLoss(alpha=1.0, beta=5.0)
optimizer = optim.Adam(model_dlinear.parameters(), lr=0.001)

epochs = 10

print("Starting DLinear Training...")
for epoch in range(epochs):
    model_dlinear.train()
    epoch_loss = 0.0

    for batch_x, batch_y in dataloader:
        optimizer.zero_grad()
        predictions = model_dlinear(batch_x)
        loss = criterion(predictions, batch_y)
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item()

    avg_loss = epoch_loss / len(dataloader)
    print(f"Epoch [{epoch+1}/{epochs}] | Average Dual-Penalty Loss: {avg_loss:.4f}")

print("Training Complete.")
```

```python
import matplotlib.pyplot as plt

model_dlinear.eval()

with torch.no_grad():
    sample_x, sample_y = next(iter(dataloader))
    pred_y = model_dlinear(sample_x)

idx = 0
hist_scaled = sample_x[idx].numpy()
true_scaled = sample_y[idx].numpy()
pred_scaled = pred_y[idx].numpy()

hist_unscaled = scaler.inverse_transform(hist_scaled)
true_unscaled = scaler.inverse_transform(true_scaled)
pred_unscaled = scaler.inverse_transform(pred_scaled)

time_hist = np.arange(-seq_length, 0)
time_pred = np.arange(0, pred_length)

plt.figure(figsize=(10, 5))
plt.plot(time_hist, hist_unscaled, label='Historical Data', color='blue')
plt.plot(time_pred, true_unscaled, label='Ground Truth', color='green', marker='o')
plt.plot(time_pred, pred_unscaled, label='DLinear Forecast', color='red', linestyle='--', marker='x')

plt.axvline(0, color='gray', linestyle='--')
plt.axhline(180, color='orange', linestyle=':', label='Hyperglycemic Threshold')
plt.axhline(70, color='red', linestyle=':', label='Hypoglycemic Threshold')

plt.title('DLinear vs. Ground Truth: Physiological Spikes')
plt.xlabel('Time Steps (5-min intervals)')
plt.ylabel('Blood Glucose (mg/dL)')
plt.legend()
plt.grid(True)
plt.tight_layout()

plt.savefig('figure1.png', dpi=300)
plt.show()
```
